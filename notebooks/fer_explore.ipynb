```python
emotions=[]
counts=[]
for emotion in os.listdir(train_path):
    emotions.append(emotion)
    counts.append(len(os.listdir(os.path.join(train_path,emotion))))
plt.bar(emotions,counts)
plt.title("Class distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
import random
from PIL  import Image
def show_random_images(emotion,n=3):
    folder=os.path.join(train_path,emotion)
    images=os.listdir(folder)
    for i in range(n):
        img_name=random.choice(images)
        img_path=os.path.join(folder,img_name)
        img=Image.open(img_path).convert('L')
        plt.imshow(img,cmap='gray')
        plt.title(emotion)
        plt.axis('off')
        plt.show()
show_random_images("happy")
show_random_images("sad")
```

```python
sizes=[]
for emotion in os.listdir(train_path):
    folder=os.path.join(train_path,emotion)
    for img_name in os.listdir(folder)[:50]:
        img_path=os.path.join(folder,img_name)
        img=Image.open(img_path)
        sizes.append(img.size)

print(sizes)
        
```

```python
count=0
for emotion in os.listdir(train_path):
    folder=os.path.join(train_path,emotion)
    for img_name in os.listdir(folder):
        img_path=os.path.join(folder,img_name)
        try:
            img=Image.oprn(img_path)
            img.verify()
        except:
            count+=1
            print(f"corrupted image:",img_path)

print(count)
```
